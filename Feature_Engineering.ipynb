```python
import pandas as pd 
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
```

```python
df= pd.read_csv("Titanic.csv")
df.columns
```

```python
numerical_coln = ['Age','Fare']
categorial_coln = ['Embarked','Sex']
```

```python
imputer = SimpleImputer(strategy = 'mean')
df[numerical_coln] = imputer.fit_transform(df[numerical_coln])
```

```python
imputer = SimpleImputer(strategy = 'most_frequent')
df[categorial_coln] = imputer.fit_transform(df[categorial_coln])
```

```python
encoder = OneHotEncoder(sparse_output = False, handle_unknown = "ignore")
encoded = encoder.fit_transform(df[categorial_coln])
```

```python
scaler = StandardScaler()
scaled = scaler.fit_transform(df[numerical_coln])
```

```python
encoded_df = pd.DataFrame(
    scaled, columns = numerical_coln)
```

```python
scaled_df = pd.DataFrame(
    scaled,
    columns=numerical_coln
)
```

```python
final_df = pd.concat([scaled_df, encoded_df], axis=1)

print(final_df.head())
```
